```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import numpy as np

# Ground truth and predictions from our classifier run
y_true = ["true"] * 289 + ["true"] * 180 + ["false"] * 4 + ["false"] * 52
y_pred = ["true"] * 289 + ["false"] * 180 + ["false"] * 4 + ["true"] * 52
```

```python
conf_mat = confusion_matrix(y_true, y_pred, labels=["true", "false"])
plt.figure(figsize=(8,6))
sns.heatmap(conf_mat, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Predicted Attack", "Predicted Legitimate"],
            yticklabels=["Actual Attack", "Actual Legitimate"])
plt.title("ClinicalShield AI — LLM Classifier Confusion Matrix")
plt.tight_layout()
plt.show()
```
